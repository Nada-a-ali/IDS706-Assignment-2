Basic Filtering and Grouping 

```python
#filtering by coffee type Arabica (below umbrella column "species" in the following dataset) 
Arabica_type = cq[cq["Species"] == "Arabica"]
len(Arabica_type)
```

```python
#filtering by coffee type Robusta (below umbrella column "species" in the following dataset)
Robusta_type = cq[cq["Species"] == "Robusta"]
len(Robusta_type)
```

```python
#using the average coffee cup quality score in the dataset, filtered by coffee cups scored above average  
average_quality = cq["Total.Cup.Points"].mean()

coffee_quality_above_average = cq[cq["Total.Cup.Points"] > average_quality]

len(coffee_quality_above_average)

coffee_quality_above_average.head()
```

```python
#Arabica type coffee cup(s) scored above average
above_q_avg_arabica_type = coffee_quality_above_average[coffee_quality_above_average["Species"] == "Arabica"]

len(above_q_avg_arabica_type)
```

```python
#Above avg. Arabica type coffee cup(s) by country of origin 
origin_country_quality_arabica = (above_q_avg_arabica_type.groupby("Country.of.Origin")
 .size().
 sort_values(ascending = False)
 )

origin_country_quality_arabica
```

```python
#Robusta type coffee cup(s) scored above average

above_q_avg_robusta_type = coffee_quality_above_average[coffee_quality_above_average["Species"] == "Robusta"]

len(above_q_avg_robusta_type)
```

```python
#Above avg. Robusta type coffee cup(s) by country of origin 

origin_country_quality_robusta = (above_q_avg_robusta_type.groupby("Country.of.Origin")
 .size().
 sort_values(ascending = False)
 )

origin_country_quality_robusta
```

```python
#Visualization of above average Arabica coffee type by country 
origin_country_quality_arabica.plot(kind = "bar", 
                                    title = "Arabica Coffee Cups Rated Above Average by Country of Origin", 
                                    color = "black", 
                                    ylabel = "Number of Coffee Cups",
                                    xlabel = "Country of Origin"
                                    )
```

```python
#Visualization of above average Robusta coffee type by country 
origin_country_quality_robusta.plot(kind = "bar", 
                                    title = "Robusta Coffee Cups Rated Above Average by Country of Origin", 
                                    color = "blue", 
                                    xlabel = "Country of Origin", 
                                    ylabel = "Number of Coffee Cups"
                                    )
```

Exploring a Machine Learning Algorithm - Basic Regression 

```python
#Import library for ML linear regression exploration 
%pip install scikit-learn
from sklearn.linear_model import LinearRegression
```

```python
#Exploring whether "sweetness" can be used to predict coffee quality via "Total.Cup.Points" 
x = cq[["Sweetness"]]
y = cq["Total.Cup.Points"]

model = LinearRegression()
model.fit(x,y)

predictions = model.predict(x)

print(predictions[:15])
```

```python
#Exploring whether "flavor" and "aftertaste" can be used to predict coffee quality via "Total.Cup.Points" 
x = cq[["Flavor", "Aftertaste"]]
y = cq["Total.Cup.Points"]

model = LinearRegression()
model.fit(x,y)

predictions = model.predict(x)

print(predictions[:15])
```
